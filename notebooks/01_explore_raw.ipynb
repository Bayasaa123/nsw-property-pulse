```python
from pyspark.sql import functions as F

LANDING = "/Volumes/workspace/nsw_property/landing/sales/"

raw = (
    spark.read.text(LANDING + "*.DAT")
    .withColumn("_source_file", F.col("_metadata.file_path"))
)

display(raw.limit(10))
```

```python
records = raw.withColumn("record_type", F.substring("value", 1, 1))

display(records.groupBy("record_type").count().orderBy("record_type"))
```

```python
records = records.withColumn("fields", F.split("value", ";"))

display(
    records.groupBy("record_type")
    .agg(F.max(F.size("fields")).alias("n_fields"))
    .orderBy("record_type")
)
```

```python
display(records.where("record_type = 'Z'").select("value"))
```

```python
display(
    records
    .where(F.col("value").contains(";6592;4;"))
    .select("record_type", "value")
)
```
